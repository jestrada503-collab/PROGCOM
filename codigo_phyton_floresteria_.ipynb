```python
estado=True
opcion=0
comp=[]
pagar=0
rta=""

print("Bienvenido a mi floristería.")
while estado:
  print("***MENÚ***")
  print("¿Qué deseas hacer?")
  print("1. Visualizar nuestras flores")
  print("2. Comprar")
  print("3. Pagar")
  print("4. Factura de compra")
  print("5. Salir")
  opcion=input("Ingrese su opción: ")
  match opcion:
    case "1":
      print("\nEstas son nuestras flores con sus respectivas características:\n")
      for f in range(len(flores["tipo"])):
        tipo = flores["tipo"][f]
        cantidad = flores["cantidad"][f]
        precio = flores["precio"][f]
        print(f"{tipo}: {cantidad} disponible, valor por unidad: {precio}\n")
    case "2":
      print("\n¿Qué flores desea comprar?")
      florac = input("Ingrese el nombre de la flor: ").strip().capitalize()
      if florac in flores["tipo"]:
        n = flores["tipo"].index(florac)
        stock_disponible = flores["cantidad"][n]

        if stock_disponible > 0:
          try:
            cant_comprar = int(input(f"¿Cuántas unidades de {florac} desea comprar? (Disponibles: {stock_disponible}): "))
            if cant_comprar <= 0:
              print("\nLa cantidad debe ser mayor a 0.\n")
            elif cant_comprar <= stock_disponible:
              costo_total = cant_comprar * flores["precio"][n]
              #para actualizar el inventario
              flores["cantidad"][n] -= cant_comprar
              comp.append({"flor": florac, "cantidad": cant_comprar, "subtotal": costo_total})
              pagar += costo_total

              print(f"\nSe han agregado {cant_comprar} {florac}(s) a tu carrito\n")
            else:
              print(f"\nNo hay suficiente stock. Solo tenemos {stock_disponible} unidades.\n")
          except ValueError:
            print("\nPor favor, pida un valor válido.\n")
        else:
          print(f"\nNo quedan unidades de {florac}.\n")
      else:
        print("\nEsa flor no existe en nuestro catálogo.\n")

    case "3":
      metodo=input("¿Cuál será su método de pago? [Efectivo, Tarjeta Débito, Tarjeta Crédito, PSE]: ").strip().capitalize()
      print(f"Debe pagar {pagar+(pagar*0.19)}")
      rta=input("¿Pagar?").strip().capitalize()
      if rta=="Si":
        print("\n¡Gracias por su compra!\n")
      elif rta=="No":
        print("\n¡PAGA!\n")
      else:
        print("\nIngresa una respuesta válida.\n")


    case "4":
      if rta=="Si":
          print("***FACTURA DE COMPRA***")
          print("Productos comprados")
          print(comp['flor'])
          print(f"Cantidad comprada: {comp['cantidad']}")
          print(f"Subtotal: {comp['subtotal']}")
          print(f"Total pagado (IVA INCLUIDO): {pagar+(pagar*0.19)}")
          print(f"Método de pago: {metodo}")
      else:
        print("\n¡PAGA!\n")



    case "5":
      print("\n¡Hasta pronto!\n")
      estado=False
    case _:
      print("\nOpción no válida.\n")
```

```python
# Inventario de la floristería
flores = {
    "tipo": ["Rosa", "Tulipan", "Girasol", "Clavel"],
    "cantidad": [20, 15, 10, 12],
    "precio": [2000, 3000, 2500, 1500]
}

estado = True
opcion = 0
comp = []
pagar = 0
rta = ""
metodo = ""
pagado = False

print("Bienvenido a mi floristería.")

while estado:
    print("\n***MENÚ***")
    print("¿Qué deseas hacer?")
    print("1. Visualizar nuestras flores")
    print("2. Comprar")
    print("3. Pagar")
    print("4. Factura de compra")
    print("5. Salir")

    opcion = input("Ingrese su opción: ").strip()

    match opcion:

        case "1":
            print("\nEstas son nuestras flores con sus respectivas características:\n")

            for f in range(len(flores["tipo"])):
                tipo = flores["tipo"][f]
                cantidad = flores["cantidad"][f]
                precio = flores["precio"][f]

                print(f"{tipo}: {cantidad} disponibles, valor por unidad: ${precio}\n")

        case "2":
            if pagado:
                print("\nEsta compra ya fue pagada. No puedes agregar más productos.")
                continue

            print("\n¿Qué flores desea comprar?")
            florac = input("Ingrese el nombre de la flor: ").strip().capitalize()

            if florac in flores["tipo"]:
                n = flores["tipo"].index(florac)
                stock_disponible = flores["cantidad"][n]

                if stock_disponible > 0:
                    try:
                        cant_comprar = int(
                            input(
                                f"¿Cuántas unidades de {florac} desea comprar? "
                                f"(Disponibles: {stock_disponible}): "
                            )
                        )

                        if cant_comprar <= 0:
                            print("\nLa cantidad debe ser mayor a 0.")

                        elif cant_comprar <= stock_disponible:
                            costo_total = cant_comprar * flores["precio"][n]

                            # Actualizar el inventario
                            flores["cantidad"][n] -= cant_comprar

                            # Agregar el producto al carrito
                            comp.append({
                                "flor": florac,
                                "cantidad": cant_comprar,
                                "subtotal": costo_total
                            })

                            pagar += costo_total

                            print(
                                f"\nSe han agregado {cant_comprar} "
                                f"{florac}(s) a tu carrito."
                            )
                            print(f"Subtotal de esta compra: ${costo_total}")

                        else:
                            print(
                                f"\nNo hay suficiente stock. "
                                f"Solo tenemos {stock_disponible} unidades."
                            )

                    except ValueError:
                        print("\nPor favor, ingrese una cantidad válida.")

                else:
                    print(f"\nNo quedan unidades de {florac}.")

            else:
                print("\nEsa flor no existe en nuestro catálogo.")

        case "3":
            if not comp:
                print("\nPrimero debes comprar alguna flor.")

            elif pagado:
                print("\nEsta compra ya ha sido pagada.")

            else:
                print("\n*** MÉTODO DE PAGO ***")
                print("1. Efectivo")
                print("2. Tarjeta Débito")
                print("3. Tarjeta Crédito")
                print("4. PSE")

                metodo_opcion = input("Seleccione su método de pago: ").strip()

                metodos = {
                    "1": "Efectivo",
                    "2": "Tarjeta Débito",
                    "3": "Tarjeta Crédito",
                    "4": "PSE"
                }

                if metodo_opcion not in metodos:
                    print("\nMétodo de pago no válido.")

                else:
                    metodo = metodos[metodo_opcion]

                    iva = pagar * 0.19
                    total = pagar + iva

                    print(f"\nSubtotal: ${pagar:.2f}")
                    print(f"IVA (19%): ${iva:.2f}")
                    print(f"Total a pagar: ${total:.2f}")

                    rta = input("¿Desea confirmar el pago? (Si/No): ").strip().lower()

                    if rta in ["si", "sí"]:
                        pagado = True
                        print("\n¡Gracias por su compra!")
                        print("Pago confirmado.")

                    elif rta == "no":
                        print("\nPago cancelado. Puedes intentarlo nuevamente.")

                    else:
                        print("\nIngresa una respuesta válida.")

        case "4":
            if not pagado:
                print("\n¡Primero debes completar el pago!")

            else:
                print("\n*** FACTURA DE COMPRA ***")
                print("Productos comprados:\n")

                for producto in comp:
                    print(f"Flor: {producto['flor']}")
                    print(f"Cantidad comprada: {producto['cantidad']}")
                    print(f"Subtotal: ${producto['subtotal']:.2f}")
                    print("-------------------------")

                iva = pagar * 0.19
                total = pagar + iva

                print(f"Subtotal de la compra: ${pagar:.2f}")
                print(f"IVA incluido (19%): ${iva:.2f}")
                print(f"Total pagado: ${total:.2f}")
                print(f"Método de pago: {metodo}")
                print("\n¡Gracias por comprar en nuestra floristería!")

        case "5":
            print("\n¡Hasta pronto!")
            estado = False

        case _:
            print("\nOpción no válida. Por favor, elija del 1 al 5.")
```
