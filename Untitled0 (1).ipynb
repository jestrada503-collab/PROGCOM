```python

dato = input("Ingrese un carácter o un número: ")

if dato.isdigit():
    print("Es un número.")
    print("Tiene", len(dato), "cifra(s).")

elif dato.isalpha():

    if len(dato) == 1 and dato.lower() in "aeiou":
        print("Es una vocal.")
    else:
        print("Es una letra.")

    if dato.isupper():
        print("Es mayúscula.")
    else:
        print("Es minúscula.")

elif dato.isalnum():
    print("Es un carácter alfanumérico.")

else:
    print("Es un carácter especial.")
```

```python
# Calculadora

# Solicitar los números
num1 = float(input("Ingrese un número: "))
num2 = float(input("Ingrese un número: "))

print("\nSeleccione una operación:")
print("1) Sumar")
print("2) Restar")
print("3) Multiplicar")
print("4) Dividir")

opcion = input("escoja la opción: ")

if opcion == "1":
    resultado = num1 + num2
    print(f"Resultado: {num1} + {num2} = {resultado}")

elif opcion == "2":
    resultado = num1 - num2
    print(f"Resultado: {num1} - {num2} = {resultado}")

elif opcion == "3":
    resultado = num1 * num2
    print(f"Resultado: {num1} * {num2} = {resultado}")

elif opcion == "4":
    if num2 != 0:
        resultado = num1 / num2
        print(f"Resultado: {num1} / {num2} = {resultado}")
    else:
        print("Error: No se acepta cero.")

else:
    print("Opción inválida.")
```
