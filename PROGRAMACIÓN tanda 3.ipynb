```python
productos = {
101: ("Arroz", 3500),
102: ("Leche", 4200),
103: ("Pan", 2500),
104: ("Huevos", 18000),
105: ("Aceite", 12000),
106: ("Azúcar", 4800),
107: ("Café", 15000),
108: ("Gaseosa", 5500),
109: ("Atún", 6500),
110: ("Chocolate", 3200)
}

IVA = 0.19
subtotal = 0

print("compras")
print("Código\Producto\Precio")
for codigo, datos in productos.items():
print(f"{codigo}\t{datos[0]}\t\t${datos[1]}")

while True:
codigo = int(input("Ingrese el código del producto (0 para terminar): "))

if codigo == 0:
break

if codigo in productos:
cantidad = int(input("Ingrese la cantidad: "))
nombre = productos[codigo][0]
precio = productos[codigo][1]

total_producto = precio * cantidad
subtotal += total_producto

print(f"Agregó {cantidad} {nombre}(s) = ${total_producto}")
else:
print("Código no válido.")

# Cálculos finales
valor_iva = subtotal * IVA
total = subtotal + valor_iva

print("recibo")
print(f"Subtotal : ${subtotal:,.0f}")
print(f"IVA (19%): ${valor_iva:,.0f}")
print(f"TOTAL : ${total:,.0f}")

# Pago
while True:
recibido = float(input("Ingrese el valor recibido: $"))

if recibido >= total:
cambio = recibido - total
print(f"vueltos: ${cambio:,.0f}")
break
else:
print("fondos insuficiente.")

print("Gracias por su compra!")
```
